```python
# IIT Patna_ Programming and data structure
# function and Recurssion

# sum of the first n Natural Number

def sum(n):
  if n==0: #base case
    return 0
  else:
    return (n+sum(n-1))

sum(11)
```

```python
#fibonacci series  0, 1, 2, 3, 5, 8, 13, 21....
def fibonacci(n):
  if n==0 or n==1:
    return n
  else:
    return (fibonacci(n-1)+fibonacci(n-2))

fibonacci(5)
```

 encapsulation (data hiding)
it means wrapping data (variables) and methods(function) into a single unit called a class and restricting direct access to data.
objective:
1.Data should not be freely accessible.
2.Data should be accessed only through functions
3.This protects data from wrong or accidental modification.

abstraction example:-

from abc import(ABC), Abstraction method

```python
from abc import ABC, abstractmethod

class vehicle(ABC):  #abstract base class
  @abstractmethod
  def start(self):
    pass

class bike (vehicle):
  def start(self):
    print("bike start with kick")


class car(vehicle):
  def start(self):
    print("car start with key")
```
