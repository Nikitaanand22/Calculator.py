```python
from textblob import TextBlob
```

```python
text=input("Enter the statement to be checked: ")

blob=TextBlob(text)

polarity=blob.sentiment.polarity



print("statement checked: ",text)
print("Polarity Score: " ,polarity)


if polarity>0:
  print("Statement is Positive")
elif polarity<0:
  print("Statement is Negative")
else:
  print("Statement is Neutral")
```

```python
from transformers import pipeline
```

6 emotions (joy, fear, disgust, anger, surprise, sadness

```python
classifier=pipeline("text-classification", model="bhadresh-savani/distilbert-base-uncased-emotion")
```

```python
text=input("Enter the text to evaluate: ")
result=classifier(text)


emotion=result[0]['label']
confidence=round(result[0]['score']*100,2)

print("statement: ",text)
print("Detected Emotion: ",emotion.upper())
print(f"Accuracy is {confidence}% for sure")
```
